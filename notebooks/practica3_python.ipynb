## Ejercicio 2 — Inventario multialmacén con listas anidadas

```python
inventario = [
    ["ALM-NORTE", "Robot Aspirador", 34, 201.50],
    ["ALM-NORTE", "Monitor 27", 58, 142.00],
    ["ALM-SUR",   "Silla Ergonomica", 12, 178.90],
    ["ALM-SUR",   "Robot Aspirador", 7, 201.50],
    ["ALM-ESTE",  "Cafetera Expres", 41, 131.20],
    ["ALM-ESTE",  "Monitor 27", 25, 142.00],
]

print(inventario[3][1])

valor_stock = []

# Calculamos el valor de cada fila porque un producto puede estar en varios almacenes.
for fila in inventario:
    producto = fila[1]
    valor = round(fila[2] * fila[3], 2)
    valor_stock.append([producto, valor])

print(valor_stock)

# Sumamos el valor de todas las filas para obtener el total del inventario.
total = 0

for elemento in valor_stock:
    total = total + elemento[1]

print("Valor total:", round(total, 2))

inventario_sur = inventario[2:4]
print(inventario_sur)

total_sur = 0

for fila in inventario_sur:
    total_sur = total_sur + fila[2] * fila[3]

print("Valor de ALM-SUR:", round(total_sur, 2))
```

### ¿Qué problema aparece si dos almacenes dejan de estar contiguos?
Que podemos dejar fuera parte de su inventario o podríamos influir el otro inventario.

## Ejercicio 3 — Manipulación destructiva y copias de listas

```python
precios_originales = [59.90, 149.00, 219.00, 18.50, 189.00, 279.00]

# 1. La asignación comparte la lista, por eso el cambio afecta a ambas variables.
precios_rebajados = precios_originales
precios_rebajados[0] = 49.90

print("Originales tras la asignación:", precios_originales)
print("Rebajados:", precios_rebajados)

# 2. Restauramos el precio antes de crear las copias independientes.
precios_originales[0] = 59.90

copia_lista = list(precios_originales)
copia_slicing = precios_originales[:]

copia_lista[0] = 39.90
copia_slicing[0] = 29.90

print("Originales sin cambios:", precios_originales)
print("Copia con list():", copia_lista)
print("Copia con slicing:", copia_slicing)

#3 Recuperamos el precio para aplicar la rebaja sobre los valores iniciales
copia_lista[0] = precios_originales[0]

for i in range(len(copia_lista)):
    copia_lista[i] = round(copia_lista[i] * 0.90, 2)

print("Copia con rebaja del 10 %:", copia_lista)

#4 Añadimos precios de ejemplo porque esta lista solo contiene precios
copia_lista.extend([45.00, 75.00])
copia_lista.insert(2, 25.00)

print("Tras añadir los tres productos:", copia_lista)

#5 Guardamos el último precio para conservar el dato que retiramos.
del copia_lista[4]
descartado = copia_lista.pop()

print("Tras eliminar los dos elementos:", copia_lista)
print("Precio descartado con pop():", descartado)

#6 Comprobamos que las copias son objetos distintos del original
print("ID del original:", id(precios_originales))
print("ID de la copia con list():", id(copia_lista))
print("ID de la copia con slicing:", id(copia_slicing))

print("Original distinto de copia_lista:",
        id(precios_originales) != id(copia_lista))
print("Original distinto de copia_slicing:",
        id(precios_originales) != id(copia_slicing))
print("Copias distintas entre sí:",
        id(copia_lista) != id(copia_slicing))

print("Originales al terminar:", precios_originales)
```

Al asignar una lista a otra variable, ambas apuntan a la misma lista y el cambio aparece en las dos. Con `list()` y `[:]` creamos copias independientes y comprobamos que sus cambios no afectan al original.

## Ejercicio 5 — Funciones propias para el cálculo de importes

```python
def calcular_importe(unidades, precio_unitario, descuento_pct=0):
    """Calcula el importe con descuento y lo redondea a dos decimales."""
    bruto = unidades * precio_unitario
    descuento = bruto * descuento_pct / 100
    return round(bruto - descuento, 2)


def resumen_pedido(unidades, precio_unitario, descuento_pct=0):
    """Devuelve una tupla con el importe bruto, el ahorro y el importe neto."""
    importe_bruto = unidades * precio_unitario
    ahorro = importe_bruto * descuento_pct / 100
    # Reutilizamos la función para aplicar el mismo cálculo del neto.
    importe_neto = calcular_importe(unidades, precio_unitario, descuento_pct)
    return round(importe_bruto, 2), round(ahorro, 2), importe_neto


print("Importe de comprobación:", calcular_importe(3, 59.90, 10))
print("Importe sin descuento:", calcular_importe(3, 59.90))

# Indicamos los nombres de los parámetros y recogemos los tres resultados.
bruto, ahorro, neto = resumen_pedido(
    unidades=3, precio_unitario=59.90, descuento_pct=10
)
print("Importe bruto:", bruto)
print("Ahorro:", ahorro)
print("Importe neto:", neto)

help(calcular_importe)

pedidos = [
    (3, 59.90, 10),
    (1, 279.00, 0),
    (5, 18.50, 20),
    (2, 149.00, 5),
    (4, 89.90, 15),
]

importes = []
for unidades, precio, descuento in pedidos:
    importe = calcular_importe(unidades, precio, descuento)
    importes.append(importe)

print("Importes:", importes)
print("Importe máximo:", max(importes))
print("Importe mínimo:", min(importes))
print("Importes ordenados:", sorted(importes))
print("Número de pedidos:", len(importes))
```

## Ejercicio 8 — Del cálculo con listas al cálculo vectorizado

```python
import numpy as np

unidades = [3, 1, 5, 2, 4, 6, 2, 8, 3, 1]
precios = [59.90, 279.00, 18.50, 149.00, 89.90, 27.50, 219.00, 14.90, 64.00, 239.00]

# Mostramos el error esperado sin detener los siguientes apartados.
try:
    print(unidades * precios)
except TypeError as error:
    print("Error al multiplicar listas:", error)

np_unidades = np.array(unidades)
np_precios = np.array(precios)
print("Unidades: dtype y shape:", np_unidades.dtype, np_unidades.shape)
print("Precios: dtype y shape:", np_precios.dtype, np_precios.shape)

importes = np_unidades * np_precios
print("Importes:",importes)
print("Suma total:", round(importes.sum(), 2))

# Conservamos importes sin recargo porque se utiliza en el siguiente ejercicio.
importes_con_recargo = importes * 1.04
print("Importes con recargo del 4 %:", importes_con_recargo)
print("Precios superiores a 100:", np.count_nonzero(np_precios > 100))

mixto = np.array([1, "dos", 3.0, True])
print("Array mixto:", mixto)
print("Tipo del array mixto:",mixto.dtype)


def calcular_con_bucle():
    resultado = []
    for i in range(len(unidades)):
        resultado.append(unidades[i] * precios[i])
    return resultado
```

```python
%timeit calcular_con_bucle()
%timeit np_unidades * np_precios
```


Al multiplicar las dos listas obtenemos un TypeError porque Python no permite multiplicarlas elemento a elemento.
El array mixto tiene dtype <U32: convierte los valores a texto para mantener un mismo tipo de dato.
En esta ejecución, el bucle tarda 728 nanosegundos y NumPy 967 nanosegundos, aunque esta prueba con diez elementos no permite concluir que el bucle sea siempre más rápido.


## Ejercicio 9 — Subsetting con máscaras booleanas

```python
# Reutilizamos los arrays del ejercicio 8, con los importes sin recargo.
caros = np_precios > 100
print("Máscara de precios caros:", caros)
print("Precios superiores a 100:",np_precios[caros])
print("Número de productos caros:", caros.sum())

# Las posiciones coinciden en ambos arrays, por eso sirve la misma máscara.
importes_caros = importes[caros]
print("Importes de esas líneas:", importes_caros)
print("Suma de sus importes:", round(importes_caros.sum(), 2))
print("Posiciones:", np.where(caros))

varias = np_unidades >= 2
print("Máscara de dos o más unidades:",varias)

seleccion = (np_precios > 100) & (np_unidades >= 2)
print("Importes con ambas condiciones:", importes[seleccion])
```

## Ejercicio 10 — Matriz de ventas por trimestre y región

```python
import numpy as np

# Filas: T1, T2, T3. Columnas: Norte, Sur, Este, Oeste, Centro.
ventas_2d = np.array([
    [15200.50, 11800.25, 12950.00, 9870.40, 13400.10],
    [16840.75, 12310.60, 13480.90, 11020.35, 14100.80],
    [12535.72, 10663.06, 11221.59, 12673.08, 12334.38],
])

print("Forma:", ventas_2d.shape)
print("Dimensiones:",ventas_2d.ndim)
print("Número de valores:", ventas_2d.size)

print("Segundo trimestre:", ventas_2d[1, :])
print("Región Centro:", ventas_2d[:, 4])
print("T3 en la región Este:", ventas_2d[2, 2])
print("T1 y T2, tres primeras regiones:")
print(ventas_2d[:2, :3])

# Sumamos las columnas de cada fila para obtener el total trimestral.
totales_trimestre = ventas_2d.sum(axis=1)
totales_region = ventas_2d.sum(axis=0)
print("Totales por trimestre:", np.round(totales_trimestre, 2))
print("Totales por región:", np.round(totales_region, 2))

# Añadimos la previsión aparte para conservar las ventas reales de los tres trimestres.
prevision = ventas_2d[2] * 1.06
ventas_con_prevision = np.vstack((ventas_2d, prevision))
print("Ventas con la cuarta fila de previsión:")
print(np.round(ventas_con_prevision, 2))

regiones = ["Norte", "Sur", "Este", "Oeste", "Centro"]
indice_mayor = np.argmax(totales_region)
print("Región con mayor facturación:",regiones[indice_mayor])
print("Facturación acumulada:", round(totales_region[indice_mayor], 2))
```

shape devuelve (3, 5), porque tenemos tres trimestres y cinco regiones. ndim devuelve 2, porque organizamos las ventas en filas y columnas. size devuelve 15, que es el número total de valores de ventas.

## Ejercicio 11 — Estadística descriptiva sobre telemetría simulada

```python
import numpy as np

# Reiniciamos el generador con la misma semilla para repetir estos datos.
rng = np.random.default_rng(2024)
temperatura = rng.normal(68, 6.5, 500)
vibracion = np.abs(rng.normal(2.4, 0.8, 500))

media = np.mean(temperatura)
mediana = np.median(temperatura)
desviacion = np.std(temperatura)

print("Media:", media)
print("Mediana:",mediana)
print("Diferencia entre media y mediana:", media - mediana)
print("Desviación típica:", desviacion)
print("Temperatura mínima:", np.min(temperatura))
print("Temperatura máxima:", np.max(temperatura))

percentiles = np.percentile(vibracion, [25, 50, 75, 95])
print("Percentiles 25, 50, 75 y 95 de vibración:", percentiles)

mediciones = np.column_stack((temperatura, vibracion))
print("Forma de la matriz:",mediciones.shape)

correlacion = np.corrcoef(temperatura, vibracion)
print("Matriz de correlación:")
print(correlacion)
print("Correlación entre temperatura y vibración:", correlacion[0, 1])

# Usamos los valores sin redondear para decidir qué lecturas superan el umbral.
umbral = media + 2 * desviacion
superan = np.count_nonzero(temperatura > umbral)
porcentaje = superan / temperatura.size * 100

print("Umbral de temperatura:", umbral)
print("Lecturas que superan el umbral:",superan)
print("Porcentaje de lecturas:", round(porcentaje, 2), "%")
```

Al fijar la semilla 2024, generamos los mismos datos al repetir el código en el mismo entorno. Así, otra persona puede reproducir el análisis y comprobar los resultados.

La media (68,03) y la mediana (68,13) son muy próximas, lo que sugiere que las temperaturas están bastante equilibradas alrededor del centro.
La correlación es aproximadamente 0,0067, muy cercana a cero, por lo que no observamos una relación lineal apreciable entre temperatura y vibración.

## Ejercicio 14 — Selección con corchetes, loc e iloc

```python
salario_serie = empleados["salario_base"]
salario_tabla = empleados[["salario_base"]]
print("Tipo con un par de corchetes:", type(salario_serie))
print("Tipo con doble corchete:",type(salario_tabla))
print(salario_serie)
print(salario_tabla)

print("\nDepartamento, ciudad y salario:")
print(empleados[["departamento", "ciudad", "salario_base"]])

# El límite final no se incluye: usamos 16 para abarcar la posición 15.
print("\nFilas en las posiciones 10 a 15:")
print(empleados[10:16])

print("\nFicha de EMP-1042:")
print(empleados.loc["EMP-1042"])

print("\nDepartamento y modalidad de los tres empleados:")
print(empleados.loc[
    ["EMP-1005", "EMP-1020", "EMP-1099"],
    ["departamento", "modalidad"]
])

print("\nTres primeras filas, columnas en posiciones 2 y 5:")
print(empleados.iloc[:3, [2, 5]])

# Los dos puntos conservan todas las filas; los nombres eligen las columnas.
print("\nNombre y porcentaje de bonus:")
print(empleados.loc[:, ["nombre", "bonus_pct"]])
```

| Forma | Selección | Ejemplo |
| --- | --- | --- |
| Corchetes `[]` | Columnas por nombre. Un slice entero selecciona filas por posición y excluye el final. | `empleados["salario_base"]` es una Serie. `empleados[["salario_base"]]` es un DataFrame. `empleados[10:16]` toma las posiciones 10 a 15. |
| `loc` | Filas y columnas por etiqueta. En un slice de etiquetas se incluyen ambos extremos cuando están presentes. | `empleados.loc["EMP-1042"]` busca ese identificador, no la posición 1042. |
| `iloc` | Filas y columnas por posición desde cero. El final del slice queda excluido. | `empleados.iloc[:3, [2, 5]]` toma las filas 0, 1 y 2 y las columnas 2 y 5. |


## Ejercicio 15 — Columnas calculadas, agregación y exportación

```python
from decimal import Decimal, ROUND_HALF_EVEN

ventas["importe_bruto"] = ventas["unidades"] * ventas["precio_unitario"]

# Calculamos con decimales exactos para no desplazar los importes de medio céntimo.
netos = []
for unidades, precio, descuento in zip(
    ventas["unidades"], ventas["precio_unitario"], ventas["descuento_pct"]
):
    bruto = Decimal(str(unidades)) * Decimal(str(precio))
    neto = bruto * (1 - Decimal(str(descuento)) / 100)
    netos.append(float(neto.quantize(Decimal("0.01"), rounding=ROUND_HALF_EVEN)))

ventas["importe_neto"] = netos
ventas["ticket_medio"] = ventas["importe_neto"] / ventas["unidades"]

print("Primeras ventas con los importes calculados:")
print(ventas[["id_pedido", "importe_bruto", "importe_neto", "ticket_medio"]].head())

facturacion_region = ventas.groupby("region")["importe_neto"].sum().sort_values(ascending=False)
print("\nFacturación por región:")
print(facturacion_region)
print("Facturación neta total:", round(ventas["importe_neto"].sum(), 2))

# Conservamos también las ventas cuyo canal todavía está vacío.
resumen_categoria_canal = ventas.groupby(["categoria", "canal"], dropna=False).agg({
    "importe_neto": "mean",
    "unidades": "sum",
})
print("\nImporte medio y unidades por categoría y canal:")
print(resumen_categoria_canal)

top_productos = ventas.groupby("producto")["importe_neto"].sum().nlargest(5)
print("\nCinco productos con mayor facturación:")
print(top_productos)

# Rellenamos los ausentes en memoria sin alterar los CSV originales.
mediana_satisfaccion = ventas["satisfaccion"].median()
ventas["satisfaccion"] = ventas["satisfaccion"].fillna(mediana_satisfaccion)
ventas["canal"] = ventas["canal"].fillna("Desconocido")
print("\nMediana usada para satisfacción:",mediana_satisfaccion)
print("Ausentes después del relleno:")
print(ventas[["satisfaccion", "canal"]].isna().sum())

# Conservamos el índice para que cada importe siga asociado a su región.
facturacion_region.to_csv("../outputs/facturacion_por_region.csv", encoding="utf-8")
print("\nArchivo guardado: ../outputs/facturacion_por_region.csv")
```

Calculamos los importes con `Decimal` y redondeamos cada pedido a dos decimales con `ROUND_HALF_EVEN`. En los empates exactos de medio céntimo elegimos el céntimo par. Después sumamos los importes redondeados. Por eso Norte suma 44.576,99 €, mientras que al sumar sin redondear cada pedido y redondear solo el total obtenemos 44.576,97 €.


## Ejercicio 16 — Operadores de comparación y booleanos

```python
import numpy as np

print(2 < 3)
print("Norte" == "norte")
print(True + True + False)
print(3 != 3.0)
print("Zapatilla" > "Mochila")

unidades = 6
precio = 149.0
descuento = 15

pedido_cumple = unidades > 4 and descuento >= 10
precio_fuera_intervalo = precio < 50 or precio > 200
print("Más de cuatro unidades y descuento de al menos 10 %:", pedido_cumple)
print("Precio inferior a 50 o superior a 200:",precio_fuera_intervalo)
print("El pedido no cumple ambas condiciones:", not pedido_cumple)

temp = np.array([64.2, 88.1, 71.0, 92.5, 69.8])
vib = np.array([1.9, 3.4, 2.1, 4.0, 2.2])

# Conservamos el error solicitado sin detener la ejecución del notebook.
try:
    print(temp > 70 and vib > 3)
except ValueError as error:
    print("Error con and:", error)

resultado_logical = np.logical_and(temp > 70, vib > 3)
resultado_operador = (temp > 70) & (vib > 3)
print("Con logical_and:", resultado_logical)
print("Con &:",resultado_operador)
print("Resultados idénticos:", np.array_equal(resultado_logical, resultado_operador))
```

`2 < 3` da `True` porque 2 es menor que 3.

`"Norte" == "norte"` da `False` porque distingue mayúsculas y minúsculas.

`True + True + False` da `2` porque `True` vale 1 y `False` vale 0.

`3 != 3.0` da `False` porque ambos tienen el mismo valor.

`"Zapatilla" > "Mochila"` da `True` porque la Z va después de la M.

`not` da `False` porque el pedido sí supera cuatro unidades y tiene al menos un 10 % de descuento.

`and` produce un `ValueError` porque intenta interpretar todo el array como un único verdadero o falso.
